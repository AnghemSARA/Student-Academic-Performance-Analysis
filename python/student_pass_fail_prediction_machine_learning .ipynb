```python
# Import libraries
import pandas as pd
import numpy as np
# Machine learning tools
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
# Evaluation metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Load the dataset
file_path = r"D:\document\AFTER_GRADUATION\DATAANALYSE\MINE\PROJECT2\student_mat_clean.csv"
df = pd.read_csv(file_path)
# Show the first five rows
df.head()
```

```python
# Check the dataset shape
df.shape
```

```python
# Check column names and data types
df.info()
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Create the target column
# If G3 is greater than or equal to 10, the student passes Otherwise, the student fails
df["result"] = df["G3"].apply(lambda grade: "Pass" if grade >= 10 else "Fail")
# Check Pass/Fail distribution
df["result"].value_counts()
```

```python
# Select features used for prediction

features = [
    "school",
    "sex",
    "age",
    "address",
    "Medu",
    "Fedu",
    "studytime",
    "failures",
    "schoolsup",
    "famsup",
    "internet",
    "absences",
    "G1",
    "G2"
]

X = df[features]
y = df["result"]

X.head()
```

```python
# Convert categorical text columns into numeric columns
X_encoded = pd.get_dummies(X, drop_first=True)
# Show the encoded dataset
X_encoded.head()
```

```python
# Check the final feature shape
X_encoded.shape
```

```python
# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training data size:", X_train.shape)
print("Testing data size:", X_test.shape)
```

```python
# Compare different machine learning models

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42)
}

for model_name, model in models.items():
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    accuracy = accuracy_score(y_test, predictions)
    print(model_name, "Accuracy:", round(accuracy * 100, 2), "%")
```

```python
# Final model:Decision Tree
# The Decision Tree was selected because it achieved the best accuracy.
final_model = DecisionTreeClassifier(max_depth=6, random_state=42)
final_model.fit(X_train, y_train)
final_predictions = final_model.predict(X_test)
final_accuracy = accuracy_score(y_test, final_predictions)
print("Final Decision Tree Accuracy:", round(final_accuracy * 100, 2), "%")
```

```python
# Detailed evaluation of the final model
print(classification_report(y_test, final_predictions))
```

```python
# Confusion matrix
confusion_matrix(y_test, final_predictions)
```

```python
# Compare actual and predicted results
comparison = pd.DataFrame({
    "Actual Result": y_test.values,
    "Predicted Result": final_predictions
})
comparison.head(10)
```
